## Introducción

¿Qué es Pandas?

Pandas es una librería de Python especializada en el análisis y manipulación de datos estructurados.

Permite trabajar con tablas, filas y columnas de forma rápida y eficiente mediante estructuras llamadas DataFrames.

Es ampliamente utilizada en proyectos de Ciencia de Datos e Inteligencia Artificial.

```
# ============================================
# EJERCICIO 1: CREACIÓN DE DATAFRAME
# ============================================

# Importamos Pandas
import pandas as pd

# Creamos un diccionario con información
datos = {
    "Nombre": ["Anyela", "Jhoana", "Helen", "Yeison"],
    "Edad": [20, 25, 30, 22],
    "Ciudad": ["Cali", "Bogotá", "Medellín", "Cartagena"]
}

# Creamos el DataFrame
df = pd.DataFrame(datos)

# Título
print("TABLA DE PERSONAS")

# Mostramos el DataFrame visualmente
display(df)
```

## Explicación

En este ejercicio se creó un DataFrame utilizando diccionarios de Python.

El DataFrame permite organizar la información en filas y columnas de manera similar a una hoja de cálculo.

```
# ============================================
# EJERCICIO 2: OPERACIONES ENTRE COLUMNAS
# ============================================

# Importamos Pandas
import pandas as pd

# Creamos los datos
datos = {
    "Empleado": ["Ana", "Luis", "Carlos"],
    "Salario_Mensual": [200000, 250000, 300000]
}

# Creamos el DataFrame
df = pd.DataFrame(datos)

# Creamos nueva columna
df["Salario_Anual"] = df["Salario_Mensual"] * 12

# Título
print("TABLA DE SALARIOS")

# Mostramos resultado visualmente
display(df)
```

## Explicación

En este ejercicio se realizaron operaciones matemáticas entre columnas para calcular el salario anual de los empleados.

```
# ============================================
# EJERCICIO 3: LECTURA DE ARCHIVOS CSV
# ============================================

# Importamos Pandas
import pandas as pd

# Creamos datos
datos = {
    "Nombre": ["Ana", "Luis", "Carlos", "María"],
    "Edad": [20, 25, 30, 22],
    "Ciudad": ["Cali", "Bogotá", "Medellín", "Buenaventura"]
}

# Creamos el DataFrame
df = pd.DataFrame(datos)

# Guardamos archivo CSV
df.to_csv("datos.csv", index=False)

# Leemos el archivo CSV
df_csv = pd.read_csv("datos.csv")

# Título
print("DATOS LEÍDOS DESDE EL ARCHIVO CSV")

# Mostramos resultados visualmente
display(df_csv)
```

## Explicación

En este ejercicio se creó un archivo CSV utilizando Pandas y posteriormente se leyó con la función read_csv().

```
# ============================================
# EJERCICIO 4: DATOS SINTÉTICOS
# ============================================

# Importamos librerías
import pandas as pd
import numpy as np

# Generamos datos aleatorios
datos = {
    "Ventas": np.random.randint(100, 1000, 10),
    "Gastos": np.random.randint(50, 500, 10)
}

# Creamos el DataFrame
df = pd.DataFrame(datos)

# Título
print("TABLA DE DATOS SINTÉTICOS")

# Mostramos resultados visualmente
display(df)
```

## Explicación

En este ejercicio se utilizaron datos sintéticos generados aleatoriamente con NumPy para simular información real.

```
# ============================================
# EJERCICIO 4: DATOS SINTÉTICOS
# ============================================

# Importamos Pandas y NumPy
import pandas as pd
import numpy as np

# Establecemos una semilla para que
# los resultados sean reproducibles
np.random.seed(10)

# Generamos datos aleatorios
datos_sinteticos = {
    "Edad": np.random.randint(18, 60, 10),
    "Salario": np.random.randint(1000000, 5000000, 10)
}

# Creamos el DataFrame con los datos generados
df_sintetico = pd.DataFrame(datos_sinteticos)

# Mostramos el DataFrame
print("Datos sintéticos generados:")
print(df_sintetico)
```

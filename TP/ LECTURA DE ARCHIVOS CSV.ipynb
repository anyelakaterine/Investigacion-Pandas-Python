```
# ============================================
# EJERCICIO 3: LECTURA DE ARCHIVOS CSV
# ============================================

# Importamos la librería Pandas
import pandas as pd

# ============================================
# CREACIÓN DE DATOS
# ============================================

datos = {
    "Nombre": ["Ana", "Luis", "Carlos", "María"],
    "Edad": [20, 25, 30, 22],
    "Ciudad": ["Cali", "Bogotá", "Medellín", "Cartagena"]
}

# ============================================
# CREAR DATAFRAME
# ============================================

df = pd.DataFrame(datos)

# ============================================
# GUARDAR ARCHIVO CSV
# ============================================

df.to_csv("datos.csv", index=False)

# ============================================
# LEER EL ARCHIVO CSV
# ============================================

df_csv = pd.read_csv("datos.csv")

# ============================================
# MOSTRAR RESULTADO
# ============================================

print(df_csv)
```

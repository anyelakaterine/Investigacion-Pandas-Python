```
# ============================================
# EJERCICIO 1: CREACIÓN DE UN DATAFRAME
# ============================================

# Importamos la librería Pandas
import pandas as pd

# Creamos un diccionario con información
# de ejemplo: nombres, edades y ciudades
datos_estudiantes = {
    "Nombre": ["Anyela", "Jhoana", "Hellen", "Yeison"],
    "Edad": [20, 25, 30, 22],
    "Ciudad": ["Cali", "Bogotá", "Medellín", "Buenaventura"]
}

# Convertimos el diccionario en un DataFrame
df_estudiantes = pd.DataFrame(datos_estudiantes)

# Mostramos el DataFrame en pantalla
print("DataFrame de estudiantes:")
print(df_estudiantes)

```

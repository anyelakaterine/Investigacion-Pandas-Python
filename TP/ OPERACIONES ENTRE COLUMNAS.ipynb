```
# ============================================
# EJERCICIO 2: OPERACIONES ENTRE COLUMNAS
# ============================================

# Importamos Pandas
import pandas as pd

# Creamos un diccionario con salarios mensuales
datos_empleados = {
    "Empleado": ["Juan", "Pedro", "Laura"],
    "Salario_Mensual": [2000000, 2500000, 3000000]
}

# Creamos el DataFrame
df_empleados = pd.DataFrame(datos_empleados)

# Creamos una nueva columna llamada
# "Salario_Anual" multiplicando el salario mensual por 12
df_empleados["Salario_Anual"] = df_empleados["Salario_Mensual"] * 12

# Mostramos el resultado
print("Cálculo de salarios anuales:")
print(df_empleados)

```
